Concatenation(+) is used to merge similar data type
 Set and Dict is not supporting Concatenation

```python
a='dk'
b='dd'

c=a+b
print(c)
```

```python
a=[1,3,4,6]
b=[2,6,8,9]

e=a+b

print(e)
```

```python
a=(1,3,4,6)
b=(2,6,8,9)

e=a+b

print(e)
```

```python
a={1,2,3,4}
b={5,6}

d= a+b

print(d)
```

```python
l=[1,3,5]

print(l[0]+2) # here it work as addition
```

```python
a={1:2,3:4}
b={5:6}

d= a+b

print(d)
```
