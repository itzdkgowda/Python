for Loop

Syntax

for item in sequence:

    # Code block to be executed for each item
    # This block must be indented

Looping through a string.

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
     print(fruit)
     
```

```python
word = "Python"
for char in word:
        print(char)
```

Looping using range() for a specific number of iterations:

```python
for i in range(5):  # Iterates from 0 to 4
    print(i)
```

```python
for i in range(20):
    print(i)
```

```python
for i in range(2,20):
    print(i)
```

```python
for i in range(1,20,2):  #(start, stop, step )
 print(i)
```

```python
for i in range(20,2,-2):  
 print(i)
```

Looping through a dictionary (keys by default):

```python
person = {"name": "Alice", "age": 30, "city": "New York"}
for key in person:
    print(key)
```

Looping through a dictionary (values)

```python
person = {"name": "Alice", "age": 30, "city": "New York"}
for value in person.values():
    print(value)
```

Looping through a dictionary (key-value pairs using items()):

```python
person = {"name": "Alice", "age": 30, "city": "New York"}
for key, value in person.items():
    print(f"{key}: {value}")
```

```python
adj = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for x in adj:
  for y in fruits:
    print(x, y)
```

for loops cannot be empty, but if you for some reason have a for loop with no content, put in the pass statement to avoid getting an error

```python
for x in [0, 1, 2]:
  pass
```

```python
for x in range(6):
    print(x)
```

```python
for x in range(6):
  if x == 3: break
  print(x)
```

```python
for x in range(6):
  if x == 3: continue
  print(x)
```

```python
message = "Hello"
for char in message:
        print(char)
```

The while Loop

With the while loop we can execute a set of statements as long as a condition is true.

```python
i = 1
while i < 6:
  print(i)
  i += 1
```

```python
i = 1
while i < 6:
  print(i)
  if i == 3:
    break
  i += 1
```

```python
i = 0
while i < 6:
  i += 1
  if i == 3:
    continue
  print(i)
```
